```python
# KMP String Matching - Manual Trace Verification
# Text    : AABCAABXAAAZ
# Pattern : AABXAA

def compute_lps(pattern):
    m = len(pattern)
    lps = [0] * m

    length = 0
    i = 1

    print("LPS Construction:")
    print("-----------------")

    while i < m:

        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length

            print(
                f"i={i}: '{pattern[i]}' == '{pattern[length-1]}' "
                f"-> LPS[{i}] = {length}"
            )

            i += 1

        else:
            if length != 0:
                length = lps[length - 1]

                print(
                    f"Mismatch at i={i}, "
                    f"fallback length -> {length}"
                )

            else:
                lps[i] = 0

                print(
                    f"i={i}: '{pattern[i]}' has no prefix match "
                    f"-> LPS[{i}] = 0"
                )

                i += 1

    return lps


def kmp_search(text, pattern):
    n = len(text)
    m = len(pattern)

    lps = compute_lps(pattern)

    print("\nLPS Array:")
    print(lps)

    print("\nKMP Matching Trace:")
    print("------------------")

    i = 0
    j = 0
    comparisons = 0
    matches = []

    while i < n:

        print(
            f"\nText index i={i}, Pattern index j={j}"
        )

        comparisons += 1

        if text[i] == pattern[j]:

            print(
                f"Compare T[{i}]='{text[i]}' "
                f"with P[{j}]='{pattern[j]}' -> MATCH"
            )

            i += 1
            j += 1

            if j == m:

                match_position = i - j
                matches.append(match_position)

                print(
                    f"Pattern found at text index "
                    f"{match_position}"
                )

                # Shift using LPS
                j = lps[j - 1]

                print(
                    f"Shift pattern using LPS -> j={j}"
                )

        else:

            print(
                f"Compare T[{i}]='{text[i]}' "
                f"with P[{j}]='{pattern[j]}' -> MISMATCH"
            )

            if j != 0:

                old_j = j
                j = lps[j - 1]

                print(
                    f"Shift using LPS: "
                    f"j={old_j} -> j={j}"
                )

            else:

                print(
                    "j=0, so move to the next text character"
                )

                i += 1

    return matches, comparisons, lps


# ------------------------------------------------
# Main Program
# ------------------------------------------------

text = "AABCAABXAAAZ"
pattern = "AABXAA"

print("========================================")
print("KMP STRING MATCHING")
print("========================================")

print("Text    :", text)
print("Pattern :", pattern)

matches, comparisons, lps = kmp_search(text, pattern)

print("\n========================================")
print("FINAL RESULT")
print("========================================")

print("LPS Array:", lps)
print("Pattern found at index:", matches)
print("Number of character comparisons:", comparisons)

if matches:
    print("Pattern exists in the text.")
else:
    print("Pattern does not exist in the text.")
```
